```python
from matplotlib import pyplot as pt
import pandas as pd
```

```python
sales_df = pd.read_excel('Sales_2025_20_Records.xlsx')
sales_df.head()
```

```python
month_wise_sales = sales_df.groupby("MonthName")['Sales_Amount'].sum()
month_wise_sales
```

```python
month = month_wise_sales.index
total_sales = month_wise_sales.values
pt.plot(month,total_sales,marker='o')
pt.ticklabel_format(axis='y',style='plain')
```

```python
admission_df = pd.read_excel('student_admission_records.xlsx')
admission_df.head(10)
```

```python
course_wise_revenue = admission_df.groupby('Course')['Total Fees'].sum()
x = course_wise_revenue.index
y = course_wise_revenue.values

pt.figure(figsize=(12, 6))
bars = pt.bar(x, y, width=0.5, color="#53df1c", edgecolor='red')

pt.title("Course-wise Revenue", fontsize=16, fontweight='bold')
pt.xlabel("Course", fontsize=12)
pt.ylabel("Total Fees", fontsize=12)

pt.xticks(rotation=20, ha='right')
pt.ticklabel_format(axis='y', style='plain')
pt.grid(axis='y', linestyle='--', alpha=0.5)
 
```

```python
remaining_fees_branch = admission_df.groupby('Branch')['Remaining Fees'].sum()
branch = remaining_fees_branch.index
total_remaining_fees = remaining_fees_branch.values

pt.barh(branch, total_remaining_fees)
```

```python
admission_df.head()
data = admission_df.groupby("Course")['Paid Fees'].sum()
course = data.index
total_paid = data.values

pt.pie(x=total_paid,labels=course,autopct='%.2f%%',explode=[0,0.1,0,0])
pt.show()
```
