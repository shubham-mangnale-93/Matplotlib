```python
from matplotlib import pyplot as plt
import pandas as pd
```

```python
sales = [10000,20000,23000,28000,35000]
year  = [2021,2022,2023,2024,2025]
plt.plot(year,sales)
plt.show()
```

```python
ad = {
    'month': ['Jan', 'Feb', 'March', 'April'],
    'no_st': [30,45,20,65]
}
student_df = pd.DataFrame(ad)
student_df
```

```python
month = student_df['month']
count = student_df['no_st']
```

```python
plt.plot(month,count,color="red",linewidth=2,linestyle='-.',marker="o",ms=10,mfc='green',mec='black')
title_font = {
    'family': 'serif',     # 'serif', 'sans-serif', 'monospace', or a specific font name
    'color':  'darkred',   # Any valid matplotlib color
    'weight': 'bold',      # 'normal', 'bold', 'light', etc.
    'size': 15,             # Numeric font size or strings like 'large', 'x-large'
}
plt.title("Month-wise Admission",loc='left',fontdict=title_font)
plt.xlabel("Month Name",loc='center',fontdict=title_font)
plt.ylabel("Nu.of Students",loc='center',fontdict=title_font)
plt.grid()
plt.show()
```
